# Mount S3 bucket as file system

This tools will mount Object Storage bucket "warehouse" as a folder in out filesystem. Execute step below to mount.
After finish run final step "umount"

```python
PROJECT_PATH='/home/huy/Documents/ICTTM/adam_datalake'
OBJ_STORAGE_ACCESS_KEY = ''
OBJ_STORAGE_SECRET_KEY = ''
```

```python
# create a settings file with your access_key:secret_key
! echo {OBJ_STORAGE_ACCESS_KEY}:{OBJ_STORAGE_SECRET_KEY} > {PROJECT_PATH}/.passwd-s3fs
! chmod 600 {PROJECT_PATH}/.passwd-s3fs
```

```python
# create mount point
!mkdir {PROJECT_PATH}/warehouse
```

```python
# mount bucket foo to ${HOME}/foo
!s3fs warehouse {PROJECT_PATH}/warehouse -o passwd_file={PROJECT_PATH}/.passwd-s3fs -o url=https://eu2.contabostorage.com -o use_path_request_style
```

```python
# access contents
!ls -la {PROJECT_PATH}/warehouse
```

```python
# umount
!umount {PROJECT_PATH}/warehouse
```
